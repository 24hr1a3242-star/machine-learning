```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.plot(x, y)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["jan", "feb", "mar", "apr", "may"]
product_a = [100, 150, 100, 170, 220]
product_b = [80, 120, 160, 200, 210]

plt.plot(months, product_a, marker="o", label="product_A")
plt.plot(months, product_b, marker="s", label="product_B")

plt.title("Product Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()

plt.savefig("sales_chart.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_a = [120, 150, 170, 140, 200, 250]
product_b = [100, 120, 150, 180, 100, 300]

plt.plot(months, product_a, marker="o", label="Product A")
plt.plot(months, product_b, marker="s", label="Product B")

plt.title("Product Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.legend()

plt.savefig("sales_chart.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["mon", "tue", "wed", "thu", "fri", "sat", "sun"]
desktop = [450, 520, 480, 500, 610, 300, 250]
mobile = [600, 620, 650, 700, 850, 900, 950]

plt.plot(months, desktop, marker="o", label="Desktop")
plt.plot(months, mobile, marker="s", label="Mobile")

plt.title("Weekly Website Visits")
plt.xlabel("Day of the Week")
plt.ylabel("Number of Visitors")

plt.grid()
plt.legend()

plt.savefig("weekly_website.png")
plt.show()
```
