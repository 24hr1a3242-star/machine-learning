```python
import pandas as pd
data1={
    "ID":[1,2,3],
    "Name":["Rahul","Sneha","Amit"],
    "Marks":[85,92,76]
}
data2={
    "ID":[4,5],
    "Name":["Priya","Rohan"],
     "Marks":[88,95]
}
df1=pd.DataFrame(data1)
df2=pd.DataFrame(data2)
print("DataFrame1:")
print(df1)
print("\nDataFrame2:")
print(df2)
df_concat=pd.concat([df1,df2],ignore_index=True)
print("\nConcatenated DataFrame:")
print(df_concat)
print("\nStudents with Marks greater than 80:")
print(df_concat[df_concat["Marks"]>80])
df_concat["Grade"]=["B","A","c","B","A"]
print("\nDataFrame after Adding Grade Column:")
print(df_concat)
```

```python
import pandas as pd
data1 = {
    "Customer": ["A", "B", "C"],
    "Basket": [70, 90, 60]
}
data2 = {
    "Customer": ["D", "E"],
    "Basket": [85, 75]
}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
print("DataFrame1:")
print(df1)
print("\nDataFrame2:")
print(df2)
df_concat = pd.concat([df1, df2], ignore_index=True)
print("\nConcatenated DataFrame:")
print(df_concat)
print("\nCustomers with Basket greater than 80:")
print(df_concat[df_concat["Basket"] > 80])
df_concat["Grade"] = ["B", "A", "C", "B", "A"]
print("\nDataFrame after Adding Grade column:")
print(df_concat)
```
